### 1. Define a Sample Form Schema

This schema will guide our AI model on what information to look for in the document. Each field specifies its `type`, `description` (for LLM guidance), and whether it's `required`.

````python
import json

# Define a sample form schema
form_schema = {
    "applicant_name": {
        "type": "single-line text",
        "description": "Full name of the applicant",
        "required": True
    },
    "date_of_birth": {
        "type": "date",
        "description": "Applicant's date of birth (e.g., YYYY-MM-DD)",
        "required": False
    },
    "street_address": {
        "type": "multi-line text",
        "description": "Applicant's full street address, including apartment number if applicable",
        "required": True
    },
    "city": {
        "type": "single-line text",
        "description": "City of residence",
        "required": True
    },
    "state": {
        "type": "dropdown",
        "description": "State of residence (two-letter abbreviation)",
        "required": True
    },
    "zip_code": {
        "type": "number",
        "description": "Zip code of residence",
        "required": True
    },
    "email_address": {
        "type": "single-line text",
        "description": "Applicant's email address",
        "required": False
    },
    "has_criminal_record": {
        "type": "checkbox",
        "description": "Does the applicant have a criminal record? Respond with 'Yes' or 'No'.",
        "required": True
    }
}

print("Form Schema Defined:")
print(json.dumps(form_schema, indent=2))
````

### 3. AI-Based Data Extraction using Gemini API

To use the Gemini API, you'll need an API key. If you don't already have one, create a key in Google AI Studio.
In Colab, add the key to the secrets manager under the "🔑" in the left panel. Give it the name `GOOGLE_API_KEY`. Then pass the key to the SDK.

We will instruct the Gemini model to extract information from the `document_content` based on the `form_schema` and return the results in a JSON format. If a field is not found or is ambiguous, the model should leave it blank.

````python
# Import the Python SDK
import google.generativeai as genai
# Used to securely store your API key
from google.colab import userdata

# Configure the Gemini API with your API key
# Replace 'YOUR_API_KEY' with your actual key or load from userdata secrets
GOOGLE_API_KEY=userdata.get('GOOGLE_API_KEY')
genai.configure(api_key=GOOGLE_API_KEY)

print("Gemini API configured.")
````

````python
# Initialize the Gemini model
gemini_model = genai.GenerativeModel('gemini-1.5-flash') # Using a suitable model for text generation

print("Gemini model initialized.")
````

````python
import re

def extract_data_with_gemini(document_text, schema):
    schema_str = json.dumps(schema, indent=2)

    prompt = f"""
You are an expert at extracting information from documents according to a specified schema.
Extract all relevant information from the document based on the JSON schema below.
For each field, extract the corresponding value. If a value is not found or is ambiguous, use an empty string.
Respect the specified field types. Output MUST be a raw JSON object only — no markdown, no code fences.

Document Text:
{document_text}

JSON Schema:
{schema_str}

Extracted Data (JSON only):
"""

    response = gemini_model.generate_content(prompt)
    raw = response.text.strip()

    # Fix 1: strip markdown code fences (```json ... ``` or ``` ... ```)
    raw = re.sub(r"^```(?:json)?\s*", "", raw)
    raw = re.sub(r"\s*```$", "", raw)
    raw = raw.strip()

    # Fix 2: extract the first {...} block if there's surrounding text
    match = re.search(r"\{.*\}", raw, re.DOTALL)
    if match:
        raw = match.group(0)

    try:
        return json.loads(raw)
    except json.JSONDecodeError as e:
        print(f"Error: Could not parse JSON — {e}")
        print("Cleaned response was:\n", raw)
        # Fix 3: return empty dict per field instead of bare {}
        return {field: "" for field in schema.keys()}
````

### 4. Present Extracted Data and Identify Missing/Required Fields

This step simulates the review process where extracted values are displayed, and any missing required fields are highlighted.

````python
import json

def review_form_data(form_schema: dict, extracted_data: dict):
    """
    Reviews extracted form data against schema and identifies missing required fields.
    """
    print("\n" + "="*50)
    print("          FORM REVIEW")
    print("="*50)

    final_form_data = {}
    missing_required_fields = []

    print("\n--- Extracted Field Details ---\n")

    for field_name, field_props in form_schema.items():
        # Get extracted value or default to empty string
        extracted_value = extracted_data.get(field_name, "")

        # Store in final data
        final_form_data[field_name] = extracted_value

        # Check if required and missing
        is_required = field_props.get('required', False)
        field_type = field_props.get('type', 'text')

        if is_required and not extracted_value:
            missing_required_fields.append(field_name)
            status = "**MISSING (Required)**"
            print(f"❌ {field_name:25} | Type: {field_type:10} | Required: Yes | {status}")
        else:
            status = "✓ Filled" if extracted_value else "Empty (Optional)"
            print(f"{'✅' if extracted_value else '⚠️'} {field_name:25} | Type: {field_type:10} | Required: {str(is_required):5} | Value: '{extracted_value}'")

    # Summary
    print("\n" + "="*50)
    print("                  SUMMARY")
    print("="*50)

    if missing_required_fields:
        print(f"❌ {len(missing_required_fields)} Required field(s) are missing:")
        for field in missing_required_fields:
            print(f"   • {field}")
        print("\n⚠️  User must manually fill these fields before saving.")
    else:
        print("✅ All required fields have been successfully extracted!")

    print(f"\nTotal fields processed: {len(form_schema)}")
    print(f"Fields with data: {len([v for v in final_form_data.values() if v])}")

    # Final Data
    print("\n" + "="*50)
    print("           FINAL FORM DATA (Ready to Save)")
    print("="*50)
    print(json.dumps(final_form_data, indent=2))

    return final_form_data, missing_required_fields


# ========================
# Example Usage (Test Data)
# ========================

if __name__ == "__main__":
    # Sample Form Schema
    form_schema = {
        "full_name": {"type": "text", "required": True},
        "email": {"type": "text", "required": True},
        "phone": {"type": "text", "required": False},
        "date_of_birth": {"type": "date", "required": True},
        "address": {"type": "textarea", "required": False},
        "employment_status": {"type": "select", "required": True},
        "agree_terms": {"type": "checkbox", "required": True}
    }

    # Sample Extracted Data from AI
    extracted_data = {
        "full_name": "Sayana Surendran",
        "email": "sayana.krish@gmail.com",
        "phone": "1234567890",
        "date_of_birth": "2000-06-12",
        # "employment_status" is missing (required)
        # "agree_terms" is missing (required)
    }

    # Run Review
    final_data, missing_fields = review_form_data(form_schema, extracted_data)
````

Now, let's create a simple web interface using Streamlit to demonstrate the data extraction in a more interactive way. This will involve:

````python
%%writefile app.py
import streamlit as st
import json
import os
import google.generativeai as genai

# --- Configuration --- #
# Get API key from environment variable (for Streamlit deployment)
# If running directly in Colab without setting an env var,
# you might need to hardcode it or pass via secrets management for deployment
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")

if GOOGLE_API_KEY:
    genai.configure(api_key=GOOGLE_API_KEY)
    gemini_model = genai.GenerativeModel('gemini-1.5-flash') # Or your preferred model
else:
    st.error("GOOGLE_API_KEY not found. Please set it as an environment variable.")
    st.stop()

# --- Form Schema (hardcoded for this demo, but could be dynamic) ---
form_schema = {
    "applicant_name": {
        "type": "single-line text",
        "description": "Full name of the applicant",
        "required": True
    },
    "date_of_birth": {
        "type": "date",
        "description": "Applicant's date of birth (e.g., YYYY-MM-DD)",
        "required": False
    },
    "street_address": {
        "type": "multi-line text",
        "description": "Applicant's full street address, including apartment number if applicable",
        "required": True
    },
    "city": {
        "type": "single-line text",
        "description": "City of residence",
        "required": True
    },
    "state": {
        "type": "dropdown",
        "description": "State of residence (two-letter abbreviation)",
        "required": True
    },
    "zip_code": {
        "type": "number",
        "description": "Zip code of residence",
        "required": True
    },
    "email_address": {
        "type": "single-line text",
        "description": "Applicant's email address",
        "required": False
    },
    "has_criminal_record": {
        "type": "checkbox",
        "description": "Does the applicant have a criminal record? Respond with 'Yes' or 'No'.",
        "required": True
    }
}

# --- Data Extraction Function ---
def extract_data_with_gemini(document_text, schema):
    """
    Uses the Gemini API to extract data from a document based on a given schema.
    """
    schema_str = json.dumps(schema, indent=2)

    prompt = f"""
You are an expert at extracting information from documents according to a specified schema. Your task is to extract all relevant information from the provided document text based on the given JSON schema. For each field in the schema, extract the corresponding value from the document. If a field is not present in the document, or if the value is ambiguous, leave its value as an empty string. Ensure the extracted data respects the specified field types and descriptions. The output MUST be a JSON object with keys corresponding to the schema fields and their extracted values.

Document Text:
{document_text}

JSON Schema:
{schema_str}

Extracted Data (JSON format):
"""

    try:
        response = gemini_model.generate_content(prompt)
        extracted_json = json.loads(response.text.strip())
        return extracted_json
    except Exception as e:
        st.error(f"Error during AI extraction: {e}")
        return {}

# --- Streamlit UI ---
st.set_page_config(layout="wide")
st.title("AI-Powered Document Data Extractor")
st.markdown("Upload document text and extract structured data using the Gemini API.")

col1, col2 = st.columns(2)

with col1:
    st.header("Document Input")
    document_content_input = st.text_area(
        "Paste your document text here:",
        height=300,
        value="""APPLICATION FORM\n\nApplicant Full Name: Sayana Surendran\nDate of Birth: 12-06-2000\n\nAddress:\n123 Main St, Apt 4B\nSpringfield, IL 62704\n\nContact Information:\nEmail: sayana.krish@gmail.com\nPhone: 1234567890\n\nCriminal Record: No\n\nPlease ensure all fields are filled accurately.\n"""
    )

    if st.button("Extract Data"):
        st.session_state['extracted_data'] = extract_data_with_gemini(document_content_input, form_schema)

with col2:
    st.header("Form Schema")
    st.json(form_schema)

    st.header("Extracted Data")
    if 'extracted_data' in st.session_state:
        extracted_data = st.session_state['extracted_data']
        st.json(extracted_data)

        st.subheader("Review and Status")
        final_form_data = {}
        missing_required_fields = []

        for field_name, field_props in form_schema.items():
            extracted_value = extracted_data.get(field_name, "")
            final_form_data[field_name] = extracted_value

            if field_props['required'] and not extracted_value:
                missing_required_fields.append(field_name)
                st.warning(f"**MISSING (Required)**: {field_name}")
            else:
                st.write(f"**{field_name}**: {extracted_value}")

        if missing_required_fields:
            st.error(f"Please fill in the following required fields: {', '.join(missing_required_fields)}")
        else:
            st.success("All required fields have been extracted.")

        st.subheader("Full Form Data (JSON)")
        st.json(final_form_data)
    else:
        st.info("Click 'Extract Data' to see results.")
````
